## Magnetometer Optimal Sensitivity

This equation represents the optimum magnetometric sensitivity, provided that the photon shot noise from the lasers does not exceed the contribution from atomic spin-projection noise. In well-calibrated instruments, this assumption holds well

$$
\delta B_{opt} \approx \frac{1}{\gamma}\sqrt{\frac{\xi}{V \tau}}
$$

For our magnetometer, $\xi$ is the relaxation constant and can range between $~10^{-9}$ cm$^3/$s and ~$10^{-13}$ cm$^3$/s for potassium atoms. Our vapour cell had a radius of $0.5$ cm which corresponds to a 0.52 cm$^3$ volume. Then for potassium the gyromagnetic ratio is approximately $2 \pi \cdot 7$ Hz/nT (this is the ratio of the magnetic dipole moment to its angular momentum).

We compute the sensitivity irrespective of measurement time (by multiplying $\delta B_{opt}$ by $\sqrt{\tau}$ and using the conservative estimate for the relaxation constant.

```python
xi = 10**-9 * ureg.cm**3 / ureg.s
r = 0.5 * ureg.cm
V = 4/3 * np.pi * r**3
tau = 10 * ureg.s
gamma = 7e9 * 2*np.pi / ureg.s / ureg.T
sensitivity = np.abs(1/gamma) * np.sqrt(xi / V)
print(sensitivity.to(ureg.fT / ureg.Hz**0.5))
```

# Working Principle

There are many different types of optically pumped quantum magnetometers (OPQM) using different measurements schemes, bandwidths, time-resolution, atomic mediums. Our experiment is an $M_x$ style potassium magnetometer.

In our typical optically pumped quantum magnetometer (OPQM), we bias the sensor with a static (or sufficiently static) test B-field, denoted $\textbf B_0$, along the $z$ direction and another oscillating B-field, denoted $\textbf B_1$,  along the $y$ direction. Under this superposition of fields, we can express the time-evolution of the vapour cell's collective magnetic moment $\vec M$ with a semi-classical treatment of the bloch equations.

$$
    \frac{d\textbf M}{dt} = \gamma \textbf M\times \textbf B
$$
